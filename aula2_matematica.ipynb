### ➕ Os Símbolos Secretos da Programação
No nosso dia a dia, a gente usa o `x` para multiplicar e o `÷` para dividir, certo? Mas no computador, os símbolos mudam só um pouquinho para não confundir o sistema. Olha a tabelinha que você vai dominar hoje:

*   `+` ➔ **Somar** (Igualzinho ao que você aprendeu na escola!)
*   `-` ➔ **Subtrair** (Também igualzinho ao da escola!)
*   `*` ➔ **Multiplicar** (Esqueça o 'x', no computador usamos o **asterisco**!)
*   `/` ➔ **Dividir** (Esqueça o '÷', usamos a **barra inclinada**!)

### 🛒 Fazendo contas com as nossas Caixinhas
Imagine que você foi à padaria e comprou pão e café. Na programação, para saber o valor total, nós criamos uma caixinha para o preço de cada item e mandamos o Python somar tudo!

Olha como a leitura fica natural e gostosa:
*   `preco_cafe = 5.50`
*   `preco_pao = 4.00`
*   `total = preco_cafe + preco_pao`

🚨 **Dica para não errar:** Na programação, nós usamos o **ponto `.`** no lugar da vírgula para separar os centavos! Se você digitar `5,50` com vírgula, o Python vai quebrar e dar erro. Lembre-se sempre: dinheiro no computador usa ponto!

```python
# === 🚀 HORA DE FAZER A CONTA DA PADARIA! ===

# 1. Guardamos os preços reais nas nossas variáveis:
preco_cafe = 5.50
preco_pao = 4.00

# 2. O Python faz a mágica da matemática sozinho:
total = preco_cafe + preco_pao

# 3. Exibimos o resultado final usando a nossa receita de print:
print(f"O café custou R$ {preco_cafe} e o pão custou R$ {preco_pao}.")
print(f"O valor total da conta deu: R$ {total} ☕")
```
