----- 

### ⚙️ Como o Python pensa por dentro? (O segredo do True e False)
Quando você escreve uma comparação como `nota >= 6.0`, o Python não lê isso como texto. Ele faz uma pergunta matemática para o sistema que só tem duas respostas possíveis na natureza da programação:
*   **`True`** (Significa **Verdadeiro** / Sim)
*   **`False`** (Significa **Falso** / Não)

Esses dois valores são um tipo de dado especial chamado **Booleano**. 
O comando **`if`** funciona como um guarda de trânsito: ele analisa o teste. Se o resultado for **`True`**, ele abre o caminho para os comandos que estão logo abaixo dele. Se o resultado for **`False`**, ele ignora tudo o que está ali e pula direto para o comando **`else`**!

⚠️ **Regra dos Caminhos Exclusivos:** O Python nunca vai rodar o `if` e o `else` ao mesmo tempo. Eles são caminhos opostos. Se ele entrar na porta do `if`, a porta do `else` se tranca automaticamente e é totalmente ignorada pelo computador.

----- 

### 📊 1. A Tabela dos Símbolos de Comparação
Para fazer o Python gerar um `True` ou `False`, você precisa usar os símbolos corretos. Olha a tabela:

*   `>` ➔ **Maior que**
*   `<` ➔ **Menor que**
*   `>=` ➔ **Maior ou igual a**
*   `<=` ➔ **Menor ou igual a**
*   `==` ➔ **Igual de verdade!**

🚨 **ATENÇÃO PARA O ERRO MAIS COMUM DO MUNDO:** Lembra que na Aula 1 usamos um sinal de `=` sozinho para guardar coisas na caixinha (ex: `idade = 20`)? Pois é! Se você quiser **PERGUNTAR** se o valor de uma caixinha é igual a outro dentro do `if`, você é obrigado a usar **dois sinais de igual grudados `==`**! Se usar um só, o Python tenta guardar o valor em vez de comparar, e o código quebra.

----- 

### 🚪 2. O Mistério do Espaço em Branco (Blocos de Código)
Você já deve ter reparado que os comandos dentro do `if` e do `else` ficam 'empurrados' para a direita. Na programação, esses 4 espaços em branco no início da linha são chamados de **Indentação**.

Pense na indentação como uma sala trancada:
*   Os dois-pontos `:` servem para abrir a porta da sala.
*   Tudo o que estiver empurrado para a direita com espaços está **dentro da sala** daquela decisão.
*   Assim que você digita um comando colado na parede esquerda de novo, você **saiu da sala**. Aquele comando vai rodar sempre, não importa o resultado do `if` ou do `else`!

----- 

### 🔢 3. Tipos de Caixinhas para Não Bugar
Para fazer perguntas que envolvem números decimais ou notas, lembre-se sempre de transformar o texto do `input()` usando o comando **`float()`**:
`nota = float(input("Digite a nota: "))`

Se você usasse o `int()`, o Python só aceitaria números inteiros redondos (como `6` ou `7`) e travaria jogando um erro de `ValueError` caso o aluno tentasse digitar uma nota quebrada como `5.9`!

```python
# === 🚀 HORA DE TESTAR O SISTEMA DE DECISÃO INTERATIVO! ===

# 1. Perguntamos a nota usando float() para aceitar números quebrados:
nota_portugues = float(input("Digite a sua nota na prova de língua portuguesa (de 0 a 10): "))

# 2. O computador analisa a nota por dentro (Gerando True ou False):
if nota_portugues >= 6.0:
    # Esta linha está dentro da sala do IF (repare o recuo à direita!)
    print("🎉 PARABÉNS! Você tirou uma nota azul e está aprovado!")
else:
    # Esta linha está dentro da sala do ELSE
    print("📚 FICA ESPERTO! Sua nota foi vermelha. Você está de recuperação!")

# Esta linha está na parede esquerda, fora das salas. Roda sempre!
print("\n[Aviso do Sistema]: A análise da nota foi concluída com sucesso.")
```
