-----

### 🔀 O que é `if` e `else`? (SE e SENÃO)
Em inglês, a palavra **`if`** significa **SE** e a palavra **`else`** significa **SENÃO**. É exatamente assim que tomamos decisões no nosso dia a dia:
*   **SE** o sinal estiver verde ➔ Eu atravesso a rua.
*   **SENÃO** (se estiver vermelho) ➔ Eu fico parado esperando.

No Python, nós usamos essa mesma lógica para fazer o computador escolher qual mensagem exibir na tela!

----- 

### ⚙️ Como o Python pensa por dentro? (O segredo do True e False)
Quando você escreve uma comparação como `nota >= 6.0`, o Python faz uma pergunta interna para o sistema que só tem duas respostas possíveis:
*   **`True`** (Significa **Verdadeiro** / Sim)
*   **`False`** (Significa **Falso** / Não)

O comando **`if`** funciona como um guarda: se o resultado for **`True`**, ele deixa rodar os comandos que estão logo abaixo dele. Se for **`False`**, ele ignora tudo e pula direto para o comando **`else`**!

⚠️ **Caminhos Opostos:** O Python nunca vai rodar o `if` e o `else` ao mesmo tempo. Se ele entrar no caminho do `if`, o caminho do `else` é totalmente ignorado!

-----

### 📊 1. A Tabela dos Símbolos de Comparação
Para fazer o Python gerar um `True` ou `False`, você usa estes símbolos:
*   `>` ➔ **Maior que**
*   `<` ➔ **Menor que**
*   `>=` ➔ **Maior ou igual a**
*   `<=` ➔ **Menor ou igual a**
*   `==` ➔ **Igual de verdade!**

🚨 **ERRO CLÁSSICO:** Lembra que na Aula 1 usamos um sinal de `=` sozinho para guardar coisas na caixinha (ex: `idade = 20`)? Se você quiser **PERGUNTAR** se algo é igual a outro dentro do `if`, use **dois sinais de igual grudados `==`**! Um só quebra o código.

-----

### 🚪 2. O Mistério do Espaço em Branco (A Margem do Código)
Você reparou que os comandos embaixo do `if` e do `else` não começam encostados no canto esquerdo? Eles ganham um recuo de **4 espaços em branco** para a direita (uma apertada na tecla Tab).

Na programação, esse espaço se chama **Indentação**, e ele funciona como uma regra de sub-itens:
*   Os **dois-pontos `:`** servem para abrir a condição e indicar que o comando da linha de baixo está vinculado ao `if` ou ao `else`.
*   Os **4 espaços** servem para travar as linhas que só devem ser executadas se aquela decisão for a escolhida.
*   Se você voltar a escrever colado no canto esquerdo, o Python entende que aquela linha está livre e **vai rodar sempre!**

-----

### 🔢 3. Tipos de Caixinhas para Não Bugar
Para fazer perguntas que envolvem notas ou dinheiro quebrados (como `5.9`), use o comando **`float()`** antes do input:
`nota = float(input("Digite a nota: "))`

Se você usasse o `int()`, o Python só aceitaria números inteiros redondos (como `6`) e travaria jogando um erro de `ValueError` na tela se o aluno tentasse digitar `5.9`!

```python
# === 🚀 HORA DE TESTAR O SISTEMA DE NOTAS INTERATIVO! ===

# 1. Perguntamos a nota usando float() para aceitar números quebrados:
nota_portugues = float(input("Digite a sua nota na prova de língua portuguesa (de 0 a 10): "))

# 2. O computador analisa a nota por dentro (Gerando True ou False):
if nota_portugues >= 6.0:
    # Esta linha está empurrada para a direita, então depende do IF!
    print("🎉 PARABÉNS! Você tirou uma nota azul e está aprovado!")
else:
    # Esta linha está empurrada para a direita, então depende do ELSE!
    print("📚 FICA ESPERTO! Sua nota foi vermelha. Você está de recuperação!")

# Esta linha voltou para o canto esquerdo, fora das decisões. Roda sempre!
print("\n[Aviso do Sistema]: A análise da nota foi concluída com sucesso.")
```
