---

### 🎤 Como funciona o `input()`?
O comando `input()` faz o computador pausar o programa, mostrar uma pergunta na tela e ficar esperando a pessoa digitar algo no teclado. Assim que a pessoa aperta Enter, o Python pega o que foi digitado e joga dentro de uma caixinha!

Olha como a leitura é simples:
`nome = input("Qual é o seu nome? ")`

---

### 🚨 O Grande Perigo do Input (Cuidado para não travar!) 💀
Preste muita atenção agora para não passar vergonha com os seus códigos. **Por padrão, o comando `input()` engole tudo o que você digita como se fosse um TEXTO (String)**, mesmo que você digite um número!

Se você pedir a idade de duas pessoas pelo input e mandar o Python somar, olha o que acontece:
*   Pessoa 1 digita: `20`
*   Pessoa 2 digita: `30`
*   O Python faz a conta: `"20" + "30"` ➔ O resultado dá **2030**! 💀 Kkkkk ele apenas grudou os textos!

🔒 **Como resolver isso?** Se você quer fazer perguntas que envolvem números inteiros (sem centavos), você precisa envelopar o input dentro do comando **`int()`**, que avisa ao computador: *"Ei, transforma essa resposta em um número de verdade!"*.

Fica assim:
`idade = int(input("Digite sua idade: "))`

```python
# === 🚀 HORA DE CONVERSAR COM O COMPUTADOR! ===

# 1. O programa faz uma pergunta de texto normal:
nome_usuario = input("Digite o seu nome completo: ")

# 2. O programa faz uma pergunta de número (usando int() para não grudar os textos):
ano_nascimento = int(input("Em que ano você nasceu? "))

# 3. O Python calcula a idade do usuário automaticamente (estamos em 2026!):
idade_calculada = 2026 - ano_nascimento

# 4. Exibimos a resposta na tela usando a nossa receita de print:
print(f"Seja muito bem-vindo, {nome_usuario}!")
print(f"O meu sistema calculou que você tem ou vai fazer {idade_calculada} anos de idade! 🎉")
```
