### 📦 1. O que é uma Variável? (As caixinhas mágicas!)
Imagine que você está organizando o seu quarto e pegou várias caixas de papelão. 

Em uma caixa, você cola uma etiqueta escrita: **"nome"** e guarda um papel com o seu nome dentro. Em outra caixa, cola uma etiqueta escrita **"idade"** e guarda o número da sua idade.

Na programação, essas caixas são chamadas de **Variáveis**. Elas servem para guardar informações na memória do computador para a gente usar quando quiser!

Olha como é fácil criar essas caixas no Python:
*   `nome = "Tom"` ➔ *Criou a caixa 'nome' e guardou o texto "Tom" dentro. (Dica de ouro: textos sempre usam aspas!)*
*   `idade = 20` ➔ *Criou a caixa 'idade' e guardou o número 20 dentro. (Números puros não precisam de aspas!)*

---

### 🗣️ 2. Como conversar com o computador? (O comando `print`)
Para fazer o computador exibir qualquer mensagem na tela, nós usamos o comando `print()`.

*   **Para textos simples (Sem caixinhas):** Basta colocar a frase entre aspas dentro dos parênteses. 
    `print("Olá, Mundo!")`

*   **Para misturar textos com as suas Caixinhas:** É aí que entra o segredo! Você coloca a letra `f` grudada antes das aspas para avisar ao Python que vai usar caixinhas, e coloca o nome da caixa **dentro de chaves `{ }`**:
    `print(f"Minha idade é {idade}")`

*🚨 **Dica de ouro para não errar:** Se você esquecer a letra `f` antes das aspas, o computador não vai abrir a caixinha. Ele vai escrever literalmente o texto com as chaves na tela, tipo: "Minha idade é {idade}".*

```python
# === 🚀 HORA DE FAZER A MÁGICA ACONTECER! ===

# 1. Aqui nós criamos as duas caixinhas na memória do Google Colab:
nome = "Tom"
idade = 20

# 2. Agora o computador vai seguir a nossa receita e ler tudo na tela:
print(f"Meu nome é {nome} e a minha idade é {idade} anos!")
```
