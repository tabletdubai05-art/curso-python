### 📦 1. O que é uma Variável? (As caixinhas mágicas!)
Imagine que você está organizando o seu quarto e pegou várias caixas de papelão. 

Em uma caixa, você cola uma etiqueta escrita: **"nome"** e guarda um papel com o seu nome dentro. Em outra caixa, cola uma etiqueta escrita **"idade"** e guarda o número da sua idade.

Na programação, essas caixas são chamadas de **Variáveis**. Elas servem para guardar informações na memória do computador para a gente usar quando quiser!

Olha como é fácil criar essas caixas no Python:
*   `nome = "Tom"` ➔ *Criou a caixa 'nome' e guardou o texto "Tom" dentro. (Dica de ouro: textos sempre usam aspas!)*
*   `idade = 20` ➔ *Criou a caixa 'idade' e guardou o número 20 dentro. (Números puros não precisam de aspas!)*

---

### 🗣️ 2. A Receita Secreta do `print()` Sem Erros!
Agora que guardamos as coisas nas caixas, como fazemos o computador abrir a caixinha e ler para a gente em uma frase bonita? Nós usamos o comando `print()`, que serve para exibir coisas na tela.

Para misturar o seu texto normal com o que está escondido dentro das caixas, existe uma **receita de bolo fixa** que os programadores profissionais usam para nunca dar erro. Vamos montar a frase passo a passo:

1.  **O comando:** Comece escrevendo `print()` com os parênteses vazios.
2.  **O segredo do 'f':** Dentro dos parênteses, coloque a letra `f` minúscula colada com aspas duplas, assim: `print(f"")`. Essa letrinha `f` é mágica! Ela avisa ao Python: *"Ei, fica esperto que vou colocar caixinhas de variáveis aqui dentro!"*.
3.  **Escreva o texto:** Digite sua frase normal dentro das aspas: `print(f"Minha idade é ")`.
4.  **Chame a caixinha:** No lugar exato onde você quer que o número apareça, coloque o nome da sua variável **dentro de chaves `{ }`**, terminando a nossa receita: `print(f"Minha idade é {idade}")`.

🚨 **Para você não travar:** O erro mais comum de quem está começando é usar parênteses `()` ou colchetes `[]` para tentar segurar a variável. Lembre-se sempre: para abrir a caixinha dentro da frase, o Python só aceita as **chaves `{ }`**!

```python
# === 🚀 HORA DE FAZER A MÁGICA ACONTECER! ===

# 1. Aqui nós criamos as duas caixinhas na memória do Google Colab:
nome = "Tom"
idade = 20

# 2. Agora o computador vai seguir a nossa receita e ler tudo na tela:
print(f"Meu nome é {nome} e a minha idade é {idade} anos!")
```
