### 📖 O que é uma Variável?
Pense na variável como uma pequena caixinha na memória do computador. Você dá um nome para essa caixinha e guarda um valor dentro dela (pode ser um texto, um número, etc.).

### 🗣️ O comando `print()`
Usamos o comando `print()` sempre que queremos exibir uma mensagem ou o conteúdo de uma caixinha (variável) na tela.

```python
# 1. Criando as variáveis (as caixinhas na memória)
nome_professor = "Tom"
curso = "Python Academy"
ano_atual = 2026

# 2. Exibindo as informações na tela usando o print
print("Olá! Seja muito bem-vindo ao curso!")
print(f"Meu nome é {nome_professor} e você está no {curso}.")
print(f"Estamos no ano de {ano_atual} criando este site juntos!")
```
