### 🗣️ 1. O comando `print()` (Exibir na tela)
Computadores são excelentes obedecendo ordens, mas eles não adivinham nada. Se você quer que o computador mostre uma mensagem na tela, você precisa usar o comando `print()` (que significa 'imprimir' ou 'exibir' em inglês).

Tudo o que você quiser que ele escreva deve ficar **dentro dos parênteses e entre aspas**, assim:
`print("Olá, Mundo!")`

---

### 📦 2. O que é uma Variável? (As caixinhas de guardar coisas)
Imagine que você está organizando a sua casa e pegou várias caixas de papelão. Em uma caixa, você escreve na frente: **"nome"** e guarda um papel com o seu nome dentro. Em outra caixa, escreve **"idade"** e guarda o número da sua idade.

Na programação, essas caixas são chamadas de **Variáveis**. Elas servem para guardar informações na memória do computador para que a gente possa usar mais tarde.

Olhe como criamos essas caixas no Python:
*   `nome = "Tom"` (Criou a caixa chamada 'nome' e guardou o texto 'Tom' dentro)
*   `idade = 20` (Criou a caixa chamada 'idade' e guardou o número 20 dentro)

*(Nota para o leigo: Textos sempre usam aspas, mas números puros não precisam de aspas!)*

```python
# === TESTE VOCÊ MESMO AGORA ===
# Aqui embaixo estão as caixinhas de exemplo na memória:
meu_nome = "Tom"
meu_curso = "Python Academy"

# Agora, vamos pedir para o computador abrir as caixinhas e ler o que está dentro:
# Usamos a letra 'f' antes das aspas para conseguir misturar texto com as caixinhas { }
print(f"Olá! Meu nome é {meu_nome} e eu sou o criador do {meu_curso}!")
```
