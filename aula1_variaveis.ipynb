### 📦 1. O que é uma Variável? (As caixinhas de guardar coisas)
Imagine várias caixas de papelão na sua casa. Em uma caixa você escreve na frente: **"nome"** e guarda seu nome dentro. Em outra, escreve **"idade"** e guarda o número da sua idade.

Na programação, essas caixas são chamadas de **Variáveis**. Elas guardam informações na memória do computador para usarmos depois.

Olhe como criamos elas no Python:
*   `nome = "Tom"` (Textos sempre usam aspas!)
*   `idade = 20` (Números puros não precisam de aspas!)

---

### 🗣️ 2. A Receita do `print()` com Caixinhas (Sem Erros!)
Para fazer o computador mostrar uma frase misturada com o que está guardado nas suas caixas, existe uma **regra fixa**:

1. **O comando:** Escreva `print()`.
2. **O segredo do 'f':** Dentro dos parênteses, coloque a letra `f` seguida de aspas duplas: `print(f"")`. O `f` avisa ao Python que você vai usar suas caixinhas.
3. **Texto e Caixinha:** Digite o texto e, onde quiser que o valor da caixa apareça, coloque o nome dela **dentro de chaves `{ }`**.

**Receita final pronta:** `print(f"minha idade é {idade}")`

*⚠️ Erro comum: Esquecer a letra `f` antes das aspas ou usar parênteses no lugar das chaves `{ }`.*

```python
# === TESTE VOCÊ MESMO AGORA ===
# 1. Criamos as caixinhas na memória:
nome = "Tom"
idade = 20

# 2. O computador vai ler o que está dentro delas usando a nossa receita:
print(f"Meu nome é {nome} e a minha idade é {idade} anos!")
```
